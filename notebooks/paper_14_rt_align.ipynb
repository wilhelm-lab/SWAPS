```python
# Load total
result_path = "/cmnfs/proj/ORIGINS/SWAPS_exp/short_gradient/30min_3to45_7R_120min_lib_im_ref_20241002_165602_293498"
maxquant_dict = pd.read_pickle(os.path.join(result_path, "maxquant_result_ref.pkl"))
ps_act = pd.read_csv(
    os.path.join(
        result_path,
        "peak_selection",
        "exp_20241003_083433_946837",
        "pept_act_sum_ps_full_tdc_fdr_thres.csv",
    )
)
ps_act_full_df = pd.merge(
    left=ps_act, right=maxquant_dict, on=["mz_rank", "Decoy"], how="left"
)
```

```python
swaps_result.plot_intensity_distr(title=None, x_lim=(1.5, 7.5))
swaps_result.plot_intensity_corr(
    show_diag=False,
    title=None,
    contour=False,
    font_size=20,
    line_width=2,
)
swaps_result.plot_overlap_with_MQ(
    show_ref=False,
    level="precursor",
    # title="30-min Exp. with 120-min Ref. Library,  RT align",
)
```
