# RT align LOWESS

```python
result_path = "/cmnfs/proj/ORIGINS/SWAPS_exp/short_gradient/30min_3to45_7R_7min_lib_im_ref_rt_lowess_20250113_153127_110363"
maxquant_result_ref = pd.read_pickle(
    os.path.join(result_path, "maxquant_result_ref.pkl")
)
ps_act_tdc_fdr_thres = pd.read_csv(
    os.path.join(
        result_path,
        "peak_selection",
        "exp_20250114_080501_712965",
        "pept_act_sum_ps_full_tdc_fdr_thres.csv",
    )
)
ps_act_full_df = pd.merge(
    left=ps_act_tdc_fdr_thres,
    right=maxquant_result_ref,
    on=["mz_rank", "Decoy"],
    how="left",
)
```

```python
%autoreload 2
from result_analysis import result_analysis
fig_save_dir = "/cmnfs/proj/ORIGINS/SWAPS_exp/SWAPS_paper_figures/"
swaps_result = result_analysis.SWAPSResult(
    maxquant_dict=maxquant_result_ref,
    pept_act_sum_df=ps_act_tdc_fdr_thres,
    infer_intensity_col="sum_intensity",
    fdr_thres=0.2,
    log_sum_intensity_thres=2,
    save_dir=os.path.join(fig_save_dir, "revision","short_to_long", "rt_align_lowess"),
    font_size=20,
    title="30-min Exp. with 7.5-min Ref. Library"
)

swaps_result.plot_intensity_distr(
    title="30-min Exp. with 7.5-min Ref. Library", x_lim=(1.5, 7.5)
)
swaps_result.plot_intensity_corr(
    show_diag=False,
    title="30-min Exp. with 7.5-min Ref. Library",
    contour=False,
    font_size=20,
    line_width=2,
)
```
